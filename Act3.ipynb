Se estima el rendimiento mensual de una acción a partir del rendimiento mensual del mercado. Los 60 datos son simulados y reproducibles.

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

rng = np.random.default_rng(20260907)
n = 60
dataset = pd.DataFrame({
    "Fecha": pd.date_range("2021-01-31", periods=n, freq="ME"),
    "Rendimiento_mercado": rng.normal(0.007, 0.04, n)
})
dataset["Rendimiento_accion"] = (
    0.004 + 1.15 * dataset["Rendimiento_mercado"] + rng.normal(0, 0.018, n)
)
print(dataset.head().to_string(index=False))

# Conclusión: se usan 60 meses simulados; no hay datos personales ni información de una empresa real.
```

```python
X = sm.add_constant(dataset[["Rendimiento_mercado"]])
Y = dataset["Rendimiento_accion"]
modelo = sm.OLS(Y, X).fit()
print(modelo.summary())

# Conclusión: el modelo calcula intercepto, coeficiente, R2, R2 ajustado y p-values.
```

## Resultados e interpretación

- Intercepto: **0.005367**.
- Coeficiente del mercado: **1.114873**. Cuando el mercado aumenta 1 punto porcentual, el rendimiento esperado de la acción cambia aproximadamente 1.115 puntos porcentuales.
- R2: **0.8567**. El modelo explica cerca de 85.7% de la variación observada en esta muestra.
- El p-value del mercado es **3.83e-26**, menor que 0.05, por lo que la variable es estadísticamente significativa al 5%.
- La asociación no demuestra causalidad y los datos simulados no sirven para recomendar inversiones.


```python
excel = {
    "Intercepto": np.float64(0.005366966703811423),
    "Coeficiente mercado": np.float64(1.1148727290969567),
    "R2": 0.8566824676212552,
    "R2 ajustado": 0.8542114756836907,
    "p-value mercado": np.float64(3.8319664254351636e-26)
}
python = {
    "Intercepto": modelo.params["const"],
    "Coeficiente mercado": modelo.params["Rendimiento_mercado"],
    "R2": modelo.rsquared,
    "R2 ajustado": modelo.rsquared_adj,
    "p-value mercado": modelo.pvalues["Rendimiento_mercado"]
}
comparacion = pd.DataFrame({"Excel": excel, "Python": python})
comparacion["Diferencia"] = (comparacion["Excel"] - comparacion["Python"]).abs()
print(comparacion.to_string())

# Conclusión: Excel y Python producen los mismos resultados; las diferencias posibles son solo de redondeo.
```
